```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\patil\OneDrive\Documents\ai-hiring-bias-xai/data/processed/resumes_cleaned.csv")
print(df.shape)
df.head()
```

```python
df.info()
```

```python
df["hired"].value_counts()
```

```python
df["hired"].value_counts().plot(kind="bar")
plt.title("Hiring Decision Distribution")
plt.xlabel("Hired (0 = No, 1 = Yes)")
plt.ylabel("Count")
plt.show()
```

```python
gender_hiring = df.groupby("gender")["hired"].mean()
print(gender_hiring)
```

```python
gender_hiring.plot(kind="bar")
plt.title("Hiring Rate by Gender")
plt.ylabel("Hiring Rate")
plt.show()
```

```python
numeric_cols = [
    "experience_years",
    "projects_count",
    "ai_score",
    "salary_expectation"
]

df[numeric_cols].hist(bins=20, figsize=(12, 8))
plt.tight_layout()
plt.show()
```

```python
for col in numeric_cols:
    df.boxplot(column=col, by="hired")
    plt.title(f"{col} vs Hiring Decision")
    plt.suptitle("")  # remove auto title
    plt.show()
```
